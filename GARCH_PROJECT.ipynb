```python
# ╔══════════════════════════════════════════════════════════════════╗
# ║   Multi-Asset Volatility Regime Detection & Risk Dashboard       ║
# ║   GARCH · HMM · VaR/CVaR                                        ║
# ║   Optimized for Google Colab — Run cells top to bottom          ║
# ╚══════════════════════════════════════════════════════════════════╝


# ════════════════════════════════════════════════════════════════════
# CELL 1 — Install Libraries
# Run this first. Colab will ask to restart runtime — click RESTART.
# ════════════════════════════════════════════════════════════════════

!pip install arch hmmlearn yfinance plotly -q


# ════════════════════════════════════════════════════════════════════
# CELL 2 — Imports
# ════════════════════════════════════════════════════════════════════

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import plotly.graph_objects as go
import plotly.express as px
from plotly.subplots import make_subplots

import yfinance as yf
from arch import arch_model
from hmmlearn.hmm import GaussianHMM
from sklearn.preprocessing import StandardScaler
from scipy import stats
from scipy.stats import jarque_bera, kurtosis, skew

import warnings
warnings.filterwarnings("ignore")

# Plotly renders inline in Colab
import plotly.io as pio
pio.renderers.default = "colab"

print("✓ All libraries imported successfully")


# ════════════════════════════════════════════════════════════════════
# CELL 3 — Configuration (Edit these values as you like)
# ════════════════════════════════════════════════════════════════════

# Tickers available on Yahoo Finance for Indian indices
TICKERS = {
    "Nifty 50":   "^NSEI",
    "Nifty Bank": "^NSEBANK",
    "Nifty IT":   "^CNXIT",
}

START_DATE       = "2018-01-01"
END_DATE         = "2024-12-31"
PORTFOLIO_VALUE  = 1_000_000    # ₹10 Lakhs — change to your portfolio size
CONFIDENCE       = 0.95         # 95% VaR confidence level
N_REGIMES        = 3            # 2 = Bull/Bear only, 3 = Bull/Sideways/Bear
LOOKBACK_WINDOW  = 252          # Trading days in 1 year

print("✓ Configuration set")
print(f"  Period       : {START_DATE} → {END_DATE}")
print(f"  Portfolio    : ₹{PORTFOLIO_VALUE:,.0f}")
print(f"  VaR Confidence: {CONFIDENCE*100:.0f}%")
print(f"  HMM Regimes  : {N_REGIMES}")


# ════════════════════════════════════════════════════════════════════
# CELL 4 — Download Data & Compute Log Returns
#
# WHY LOG RETURNS?
#   r(t) = log(P(t)/P(t-1)) × 100
#   - Time-additive: weekly return = sum of daily log returns
#   - More symmetric / closer to normal than simple returns
#   - Industry standard in quantitative finance
# ════════════════════════════════════════════════════════════════════

prices_dict  = {}
returns_dict = {}

for name, ticker in TICKERS.items():
    print(f"Downloading {name} ({ticker})...")
    raw = yf.download(ticker, start=START_DATE, end=END_DATE, progress=False)

    # For NSE indices Yahoo Finance uses 'Close' (no Adj Close)
    price_series = raw["Close"].squeeze()
    price_series.name = name
    prices_dict[name] = price_series

    # Log return in percentage
    ret = np.log(price_series / price_series.shift(1)) * 100
    ret.dropna(inplace=True)
    returns_dict[name] = ret

# Combine into DataFrames
prices  = pd.DataFrame(prices_dict).dropna()
returns = pd.DataFrame(returns_dict).dropna()

print(f"\n✓ Data loaded")
print(f"  Trading days : {len(returns)}")
print(f"  Date range   : {returns.index[0].date()} → {returns.index[-1].date()}")


# ════════════════════════════════════════════════════════════════════
# CELL 5 — Exploratory Data Analysis (EDA)
#
# Always inspect your data before modeling.
# Key stats for financial returns:
#   Skewness < 0 → left tail (more extreme negative events)
#   Excess Kurtosis > 0 → fat tails (more extremes than normal)
#   Jarque-Bera test → tests whether distribution is Normal
#     p-value < 0.05 → reject normality (almost always true for returns)
# ════════════════════════════════════════════════════════════════════

print("═" * 60)
print("DESCRIPTIVE STATISTICS")
print("═" * 60)
print(returns.describe().round(4))

print("\n" + "═" * 60)
print("DISTRIBUTION PROPERTIES")
print("═" * 60)
print(f"{'Index':<15} {'Skewness':>10} {'Ex.Kurtosis':>13} {'JB p-value':>12}")
print("-" * 55)

for col in returns.columns:
    s   = skew(returns[col])
    k   = kurtosis(returns[col], fisher=True)   # excess kurtosis (normal=0)
    jb_stat, jb_p = jarque_bera(returns[col])
    print(f"{col:<15} {s:>10.4f} {k:>13.4f} {jb_p:>12.6f}")

print("\nInterpretation:")
print("  Skewness < 0 → left-skewed (bigger downside events)")
print("  Ex.Kurtosis > 0 → fat tails (more extreme days than Normal predicts)")
print("  JB p-value < 0.05 → returns are NOT normally distributed")


# ════════════════════════════════════════════════════════════════════
# CELL 6 — Plot Prices and Returns
# ════════════════════════════════════════════════════════════════════

fig, axes = plt.subplots(len(TICKERS), 2, figsize=(16, 4 * len(TICKERS)))
fig.suptitle("Price Series & Daily Log Returns", fontsize=14, fontweight="bold")

for i, name in enumerate(TICKERS.keys()):
    # Price
    axes[i, 0].plot(prices.index, prices[name], color="#3b82f6", linewidth=0.8)
    axes[i, 0].set_title(f"{name} — Price")
    axes[i, 0].set_ylabel("Index Level")
    axes[i, 0].grid(alpha=0.3)

    # Returns with color for positive/negative
    ret = returns[name]
    colors = ["#22c55e" if r >= 0 else "#ef4444" for r in ret]
    axes[i, 1].bar(ret.index, ret, color=colors, width=1, alpha=0.7)
    axes[i, 1].set_title(f"{name} — Log Returns (%)")
    axes[i, 1].set_ylabel("Return (%)")
    axes[i, 1].grid(alpha=0.3)
    axes[i, 1].axhline(0, color="black", linewidth=0.5)

plt.tight_layout()
plt.savefig("01_prices_returns.png", dpi=150, bbox_inches="tight")
plt.show()
print("✓ Chart saved as 01_prices_returns.png")


# ════════════════════════════════════════════════════════════════════
# CELL 7 — GARCH(1,1) Modeling
#
# GARCH(1,1) Variance Equation:
#   σ²(t) = ω + α·ε²(t-1) + β·σ²(t-1)
#
#   ω (omega) : baseline/floor variance (always positive)
#   α (alpha) : ARCH term — weight on last period's shock squared
#               High α → vol reacts quickly to news
#   β (beta)  : GARCH term — weight on last period's variance
#               High β → vol decays slowly (long memory)
#   α+β       : Persistence. Near 1 = shocks last a long time.
#               Must be < 1 for stationarity.
#
# Long-run (unconditional) volatility:
#   σ_LR = sqrt(ω / (1 - α - β))
# ════════════════════════════════════════════════════════════════════

garch_results = {}   # stores fitted model objects
vol_data      = {}   # stores volatility DataFrames

print("═" * 60)
print("GARCH(1,1) MODEL FITTING")
print("═" * 60)

for name in TICKERS.keys():
    series = returns[name]

    # Fit GARCH(1,1) with Normal innovations
    model = arch_model(
        series,
        vol="Garch",
        p=1,           # ARCH order: lags of past shocks
        q=1,           # GARCH order: lags of past variance
        dist="normal", # Innovation distribution
        mean="Constant"
    )
    res = model.fit(disp="off")   # disp="off" = no iteration output
    garch_results[name] = res

    # Extract parameters
    omega = res.params["omega"]
    alpha = res.params["alpha[1]"]
    beta  = res.params["beta[1]"]
    persistence = alpha + beta
    long_run_vol = np.sqrt(omega / (1 - persistence)) * np.sqrt(252)

    print(f"\n{name}")
    print(f"  ω (omega)     : {omega:.6f}")
    print(f"  α (alpha)     : {alpha:.4f}  ← how fast vol reacts to shocks")
    print(f"  β (beta)      : {beta:.4f}  ← how long vol persists")
    print(f"  α + β         : {persistence:.4f}  ← persistence (closer to 1 = longer memory)")
    print(f"  Long-run vol  : {long_run_vol:.2f}% (annualized)")
    print(f"  AIC           : {res.aic:.2f}")
    print(f"  BIC           : {res.bic:.2f}")

    # Store volatility series
    cond_vol   = res.conditional_volatility
    annual_vol = cond_vol * np.sqrt(252)

    vol_data[name] = pd.DataFrame({
        "returns":    series,
        "cond_vol":   cond_vol,
        "annual_vol": annual_vol
    })

print("\n✓ GARCH models fitted for all indices")


# ════════════════════════════════════════════════════════════════════
# CELL 8 — Plot GARCH Conditional Volatility
# ════════════════════════════════════════════════════════════════════

fig, axes = plt.subplots(len(TICKERS), 1, figsize=(16, 5 * len(TICKERS)))
fig.suptitle("GARCH(1,1) Conditional Volatility (Annualized %)",
             fontsize=14, fontweight="bold")

for i, name in enumerate(TICKERS.keys()):
    ax = axes[i]
    vol = vol_data[name]["annual_vol"]
    ax.fill_between(vol.index, vol, alpha=0.4, color="#f59e0b")
    ax.plot(vol.index, vol, color="#f59e0b", linewidth=0.8, label="Annualized Vol")

    # Mark high-volatility periods (above 2× median)
    threshold = vol.median() * 2
    ax.axhline(threshold, color="red", linestyle="--", linewidth=0.8,
               label=f"2× Median ({threshold:.1f}%)")

    ax.set_title(f"{name} — GARCH Volatility")
    ax.set_ylabel("Annualized Volatility (%)")
    ax.legend(loc="upper right", fontsize=8)
    ax.grid(alpha=0.3)

plt.tight_layout()
plt.savefig("02_garch_volatility.png", dpi=150, bbox_inches="tight")
plt.show()
print("✓ Chart saved as 02_garch_volatility.png")


# ════════════════════════════════════════════════════════════════════
# CELL 9 — Model Comparison (GARCH vs EGARCH vs GJR-GARCH)
#
# Lower AIC/BIC = better model (penalizes unnecessary complexity)
#
# EGARCH: Models log(σ²) — captures leverage effect
#   γ < 0 → negative shocks raise vol MORE than positive shocks
#   (Bad news hurts more than good news helps — true in markets)
#
# GJR-GARCH: Adds indicator term for negative shocks (simpler than EGARCH)
#   o=1 adds the asymmetry term (o for "order of asymmetry")
# ════════════════════════════════════════════════════════════════════

print("═" * 60)
print("MODEL COMPARISON — AIC / BIC")
print("═" * 60)

# Focus on Nifty 50 for comparison
primary = "Nifty 50"
series  = returns[primary]

model_configs = {
    "GARCH(1,1)":     dict(vol="Garch",  p=1, o=0, q=1),
    "GJR-GARCH(1,1)": dict(vol="Garch",  p=1, o=1, q=1),
    "EGARCH(1,1)":    dict(vol="EGARCH", p=1, o=1, q=1),
}

print(f"\n{'Model':<20} {'AIC':>10} {'BIC':>10}  Notes")
print("-" * 60)

comparison = {}
for label, cfg in model_configs.items():
    try:
        m = arch_model(series, mean="Constant", dist="normal", **cfg)
        r = m.fit(disp="off")
        comparison[label] = r.aic
        note = ""
        if "gamma[1]" in r.params.index:
            g = r.params["gamma[1]"]
            note = f"γ={g:.4f} {'(leverage ✓)' if g < 0 else '(no leverage)'}"
        print(f"{label:<20} {r.aic:>10.2f} {r.bic:>10.2f}  {note}")
    except Exception as e:
        print(f"{label:<20} {'ERROR':>10}  {str(e)[:30]}")

best = min(comparison, key=comparison.get)
print(f"\n✓ Best model by AIC: {best}")
print("  (We use GARCH(1,1) throughout for simplicity and interpretability)")


# ════════════════════════════════════════════════════════════════════
# CELL 10 — GARCH Volatility Forecast (Next 10 Days)
#
# GARCH forecasts mean-revert to long-run volatility.
# Short horizon → dominated by current vol level
# Long horizon  → converges to unconditional (long-run) vol
# This mean-reversion property is economically intuitive:
# extreme volatility episodes always eventually calm down.
# ════════════════════════════════════════════════════════════════════

FORECAST_HORIZON = 10
print(f"═" * 60)
print(f"GARCH VOLATILITY FORECAST — NEXT {FORECAST_HORIZON} TRADING DAYS")
print("═" * 60)

for name in TICKERS.keys():
    res = garch_results[name]
    forecasts = res.forecast(horizon=FORECAST_HORIZON, reindex=False)

    # forecast.variance returns σ²(t+h) for h=1..horizon
    forecast_var = forecasts.variance.iloc[-1]
    forecast_vol = np.sqrt(forecast_var)           # daily σ
    forecast_ann = forecast_vol * np.sqrt(252)     # annualized

    current_vol = vol_data[name]["annual_vol"].iloc[-1]
    print(f"\n{name}  (Current annualized vol: {current_vol:.2f}%)")
    print(f"  {'Day':<8} {'Daily σ':>10} {'Annualized':>12}")
    print(f"  {'-'*32}")
    for day, (d, a) in enumerate(zip(forecast_vol, forecast_ann), 1):
        print(f"  Day +{day:<3} {d:>10.3f}% {a:>11.2f}%")


# ════════════════════════════════════════════════════════════════════
# CELL 11 — HMM Regime Detection
#
# FEATURE MATRIX:
#   We feed two features into the HMM:
#   [1] Daily log return   → captures direction (bull vs bear)
#   [2] GARCH volatility   → captures turbulence (calm vs crisis)
#
# SCALING:
#   StandardScaler transforms each feature to mean=0, std=1.
#   Without scaling, one feature could dominate the HMM.
#
# FITTING:
#   EM algorithm estimates:
#     - Transition matrix A[i][j] = P(state j | was in state i)
#     - Emission params: μ_k, Σ_k for each state k
#     - Initial state probabilities π
#
# VITERBI DECODING:
#   Given the fitted model, finds the most likely
#   sequence of hidden states for the observed data.
# ════════════════════════════════════════════════════════════════════

print("═" * 60)
print(f"HMM REGIME DETECTION  ({N_REGIMES} regimes)")
print("═" * 60)

# Use Nifty 50 for regime detection
primary = "Nifty 50"
ret_series = returns[primary]
vol_series = vol_data[primary]["cond_vol"]

# Build feature matrix
aligned = pd.DataFrame({
    "returns": ret_series,
    "cond_vol": vol_series
}).dropna()

scaler = StandardScaler()
X = scaler.fit_transform(aligned[["returns", "cond_vol"]])

# Fit HMM
print("\nFitting HMM (this may take a few seconds)...")
hmm_model = GaussianHMM(
    n_components=N_REGIMES,
    covariance_type="full",   # each state gets its own covariance matrix
    n_iter=2000,
    random_state=42,
    tol=1e-4
)
hmm_model.fit(X)

# Viterbi decoding — most likely state sequence
hidden_states = hmm_model.predict(X)
log_likelihood = hmm_model.score(X)
print(f"✓ HMM fitted  |  Log-Likelihood: {log_likelihood:.2f}")

# ── Label states by mean return ──────────────────────────────────
# HMM states are just 0,1,2 — we give them economic meaning.
# Sort by mean return component of each state's Gaussian mean.
state_order = np.argsort(hmm_model.means_[:, 0])

if N_REGIMES == 2:
    label_map = {state_order[0]: "Bear 🔴",
                 state_order[1]: "Bull 🟢"}
    color_map = {"Bear 🔴": "#ef4444",
                 "Bull 🟢": "#22c55e"}
else:
    label_map = {state_order[0]: "Bear 🔴",
                 state_order[1]: "Sideways 🟡",
                 state_order[2]: "Bull 🟢"}
    color_map = {"Bear 🔴":    "#ef4444",
                 "Sideways 🟡": "#f59e0b",
                 "Bull 🟢":    "#22c55e"}

regime_labels = pd.Series(
    [label_map[s] for s in hidden_states],
    index=aligned.index
)

# ── Regime statistics ─────────────────────────────────────────────
print("\n── Regime Statistics ──")
for state_num, label in label_map.items():
    mask         = hidden_states == state_num
    state_ret    = ret_series.values[mask[:len(ret_series)]]
    state_vol    = vol_series.values[mask[:len(vol_series)]]
    n_days       = mask.sum()
    pct          = n_days / len(hidden_states) * 100

    print(f"\n{label}  ({n_days} days, {pct:.1f}% of sample)")
    print(f"  Mean daily return : {state_ret.mean():+.3f}%")
    print(f"  Daily volatility  : {state_ret.std():.3f}%")
    print(f"  Annualized vol    : {state_ret.std()*np.sqrt(252):.2f}%")
    print(f"  Worst day         : {state_ret.min():.3f}%")
    print(f"  Best day          : {state_ret.max():.3f}%")

# ── Transition matrix ─────────────────────────────────────────────
print("\n── Transition Probability Matrix ──")
print("(P[row → col]: probability of moving from row-regime to col-regime)\n")

T = hmm_model.transmat_
labels_list = [label_map[i] for i in range(N_REGIMES)]

header = f"{'':18}"
for l in labels_list:
    header += f"{l:>18}"
print(header)

for i, from_l in enumerate(labels_list):
    row = f"{from_l:18}"
    for j in range(N_REGIMES):
        row += f"{T[i][j]:>18.4f}"
    # Expected duration: E[days] = 1 / (1 - P(stay))
    stay = T[i][i]
    exp_dur = 1 / (1 - stay) if stay < 1 else float("inf")
    row += f"  → Avg duration: {exp_dur:.1f} days"
    print(row)


# ════════════════════════════════════════════════════════════════════
# CELL 12 — Plot Regimes on Price Chart
# ════════════════════════════════════════════════════════════════════

fig, axes = plt.subplots(3, 1, figsize=(16, 12), sharex=True)
fig.suptitle(f"Nifty 50 — Regime Detection (HMM, {N_REGIMES} states)",
             fontsize=14, fontweight="bold")

nifty_price = prices["Nifty 50"].reindex(aligned.index)

# ── Panel 1: Price with regime background ────────────────────────
ax1 = axes[0]
ax1.plot(nifty_price.index, nifty_price, color="#3b82f6", linewidth=0.9, zorder=3)

# Color background by regime (group consecutive same-regime days)
regime_changes = (regime_labels != regime_labels.shift()).cumsum()
for _, grp in aligned.groupby(regime_changes):
    regime = regime_labels[grp.index[0]]
    color  = color_map[regime]
    ax1.axvspan(grp.index[0], grp.index[-1],
                alpha=0.25, color=color, linewidth=0)

# Legend patches
patches = [mpatches.Patch(color=c, alpha=0.5, label=l)
           for l, c in color_map.items()]
ax1.legend(handles=patches, loc="upper left", fontsize=9)
ax1.set_title("Price with Regime Background")
ax1.set_ylabel("Index Level")
ax1.grid(alpha=0.3)

# ── Panel 2: Regime as discrete steps ────────────────────────────
ax2 = axes[1]
state_numeric = pd.Series(hidden_states, index=aligned.index).astype(float)
for state_num, label in label_map.items():
    mask_plot = state_numeric[state_numeric == state_num]
    ax2.scatter(mask_plot.index, mask_plot,
                color=color_map[label], s=2, alpha=0.8, label=label)

ax2.set_title("Hidden State Sequence (Viterbi Decoded)")
ax2.set_ylabel("State")
ax2.set_yticks(list(label_map.keys()))
ax2.set_yticklabels([label_map[k] for k in label_map.keys()], fontsize=8)
ax2.grid(alpha=0.3)

# ── Panel 3: GARCH Volatility colored by regime ──────────────────
ax3 = axes[2]
annual_vol_aligned = vol_data["Nifty 50"]["annual_vol"].reindex(aligned.index)

for state_num, label in label_map.items():
    mask = hidden_states == state_num
    dates_in_state = aligned.index[mask]
    vol_in_state   = annual_vol_aligned[dates_in_state]
    ax3.scatter(dates_in_state, vol_in_state,
                color=color_map[label], s=2, alpha=0.7, label=label)

ax3.set_title("GARCH Volatility (colored by Regime)")
ax3.set_ylabel("Annualized Vol (%)")
ax3.legend(loc="upper right", fontsize=9, markerscale=5)
ax3.grid(alpha=0.3)

plt.tight_layout()
plt.savefig("03_regime_detection.png", dpi=150, bbox_inches="tight")
plt.show()
print("✓ Chart saved as 03_regime_detection.png")


# ════════════════════════════════════════════════════════════════════
# CELL 13 — VaR and CVaR Calculation
#
# THREE METHODS:
#
# 1. Historical Simulation (non-parametric)
#    → Sort past returns, take (1-confidence) percentile
#    → No distribution assumption
#    → CVaR = average of all returns worse than VaR
#
# 2. Parametric / Normal
#    → VaR = μ - z·σ  (z = 1.645 for 95%)
#    → Assumes Normal distribution — underestimates tail risk
#    → Quick to compute, easy to explain
#
# 3. GARCH Dynamic VaR (best method)
#    → VaR(t) = μ - z·σ(t)  where σ(t) comes from GARCH
#    → Adapts to current volatility conditions
#    → Gives early warning as vol rises
#    → Industry standard at banks and hedge funds
# ════════════════════════════════════════════════════════════════════

print("═" * 60)
print(f"VAR & CVAR ANALYSIS — NIFTY 50")
print(f"Portfolio: ₹{PORTFOLIO_VALUE:,.0f}  |  Confidence: {CONFIDENCE*100:.0f}%")
print("═" * 60)

nifty_ret = returns["Nifty 50"].dropna()
nifty_vol = vol_data["Nifty 50"]["cond_vol"]
garch_res  = garch_results["Nifty 50"]
z_score    = stats.norm.ppf(1 - CONFIDENCE)   # negative (left tail)
mu         = nifty_ret.mean()

# ── Method 1: Historical VaR ──────────────────────────────────────
def hist_var_cvar(ret_arr, conf, pv):
    sorted_r   = np.sort(ret_arr)
    idx        = int((1 - conf) * len(sorted_r))
    var_pct    = sorted_r[idx]
    cvar_pct   = sorted_r[:idx].mean()
    return var_pct, cvar_pct, abs(var_pct/100)*pv, abs(cvar_pct/100)*pv

# ── Method 2: Parametric VaR ──────────────────────────────────────
def param_var_cvar(ret_series, conf, pv):
    mu_    = ret_series.mean()
    sigma_ = ret_series.std()
    z_     = stats.norm.ppf(1 - conf)
    var_pct  = mu_ + z_ * sigma_
    cvar_pct = mu_ - sigma_ * stats.norm.pdf(z_) / (1 - conf)
    return var_pct, cvar_pct, abs(var_pct/100)*pv, abs(cvar_pct/100)*pv

# ── Method 3: GARCH Dynamic VaR ──────────────────────────────────
cond_vol_series = garch_res.conditional_volatility.reindex(nifty_ret.index).dropna()
var_garch_pct   = mu + z_score * cond_vol_series
cvar_garch_pct  = mu - cond_vol_series * stats.norm.pdf(z_score) / (1 - CONFIDENCE)
var_garch_inr   = abs(var_garch_pct  / 100) * PORTFOLIO_VALUE
cvar_garch_inr  = abs(cvar_garch_pct / 100) * PORTFOLIO_VALUE

# ── Print comparison ──────────────────────────────────────────────
print(f"\n{'Method':<25} {'VaR (%)':>10} {'VaR (₹)':>14} {'CVaR (%)':>10} {'CVaR (₹)':>14}")
print("-" * 78)

for label, func, args in [
    ("Historical",  hist_var_cvar,  (nifty_ret.values, CONFIDENCE, PORTFOLIO_VALUE)),
    ("Parametric",  param_var_cvar, (nifty_ret, CONFIDENCE, PORTFOLIO_VALUE)),
]:
    vp, cp, vi, ci = func(*args)
    print(f"{label:<25} {vp:>10.3f}% {vi:>14,.0f} {cp:>10.3f}% {ci:>14,.0f}")

# GARCH — most recent value
print(f"{'GARCH Dynamic (latest)':<25} "
      f"{var_garch_pct.iloc[-1]:>10.3f}% "
      f"{var_garch_inr.iloc[-1]:>14,.0f} "
      f"{cvar_garch_pct.iloc[-1]:>10.3f}% "
      f"{cvar_garch_inr.iloc[-1]:>14,.0f}")

# ── Multi-confidence table ────────────────────────────────────────
print(f"\n── Historical VaR at Multiple Confidence Levels ──")
print(f"{'Confidence':<14} {'VaR (%)':>10} {'VaR (₹)':>14} {'CVaR (%)':>10} {'CVaR (₹)':>14}")
print("-" * 58)
for conf in [0.90, 0.95, 0.99]:
    vp, cp, vi, ci = hist_var_cvar(nifty_ret.values, conf, PORTFOLIO_VALUE)
    print(f"{conf*100:.0f}%{'':<11} {vp:>10.3f}% {vi:>14,.0f} {cp:>10.3f}% {ci:>14,.0f}")


# ════════════════════════════════════════════════════════════════════
# CELL 14 — Plot Dynamic VaR/CVaR Over Time
# ════════════════════════════════════════════════════════════════════

fig, axes = plt.subplots(2, 1, figsize=(16, 9), sharex=True)
fig.suptitle(f"GARCH Dynamic VaR & CVaR — ₹{PORTFOLIO_VALUE:,.0f} Portfolio "
             f"({CONFIDENCE*100:.0f}% Confidence)", fontsize=13, fontweight="bold")

# ── Panel 1: VaR and CVaR in ₹ ───────────────────────────────────
ax1 = axes[0]
ax1.fill_between(cvar_garch_inr.index, cvar_garch_inr,
                 alpha=0.3, color="#ef4444", label="CVaR (Expected Shortfall)")
ax1.fill_between(var_garch_inr.index, var_garch_inr,
                 alpha=0.5, color="#f59e0b", label=f"VaR ({CONFIDENCE*100:.0f}%)")
ax1.plot(var_garch_inr.index,  var_garch_inr,  color="#f59e0b", linewidth=0.8)
ax1.plot(cvar_garch_inr.index, cvar_garch_inr, color="#ef4444", linewidth=0.8)
ax1.set_ylabel("Loss Estimate (₹)")
ax1.legend(loc="upper left")
ax1.grid(alpha=0.3)
ax1.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f"₹{x:,.0f}"))
ax1.set_title("Dynamic Risk Estimates (₹) — Peak = COVID/High Stress Periods")

# ── Panel 2: Returns with VaR threshold ──────────────────────────
ax2 = axes[1]
nifty_ret_aligned = nifty_ret.reindex(var_garch_pct.index)

ax2.bar(nifty_ret_aligned.index, nifty_ret_aligned,
        color=["#22c55e" if r >= 0 else "#ef4444" for r in nifty_ret_aligned],
        width=1, alpha=0.6, label="Daily Returns")
ax2.plot(var_garch_pct.index, var_garch_pct,
         color="#f59e0b", linewidth=1, label="VaR Threshold (%)")

# Highlight VaR breaches (actual loss > VaR threshold)
breaches = nifty_ret_aligned < var_garch_pct
ax2.scatter(nifty_ret_aligned[breaches].index,
            nifty_ret_aligned[breaches],
            color="black", s=12, zorder=5, label=f"VaR Breach ({breaches.sum()} events)")

ax2.set_ylabel("Return (%)")
ax2.set_xlabel("Date")
ax2.legend(loc="lower left", fontsize=9)
ax2.grid(alpha=0.3)
ax2.set_title("Returns vs VaR Threshold (black dots = breach events)")

plt.tight_layout()
plt.savefig("04_var_cvar.png", dpi=150, bbox_inches="tight")
plt.show()
print(f"✓ Chart saved as 04_var_cvar.png")
print(f"\nVaR Breaches: {breaches.sum()} out of {len(breaches)} days "
      f"({breaches.mean()*100:.2f}% vs expected {(1-CONFIDENCE)*100:.1f}%)")


# ════════════════════════════════════════════════════════════════════
# CELL 15 — VaR Backtesting (Kupiec's POF Test)
#
# Backtesting checks if the VaR model is well-calibrated.
#
# KUPIEC's PROPORTION OF FAILURES (POF) TEST:
#   H0: Breach rate = (1 - confidence)  [model is correct]
#   H1: Breach rate ≠ (1 - confidence)  [model is mis-specified]
#
#   Test statistic follows a Chi-squared distribution (df=1)
#   p-value > 0.05 → model passes backtesting (fail to reject H0)
#   p-value < 0.05 → model fails backtesting
#
# Basel III requires banks to backtest VaR over 250 trading days.
# Too many breaches → higher capital requirements.
# ════════════════════════════════════════════════════════════════════

print("═" * 60)
print("VAR BACKTESTING — KUPIEC'S POF TEST")
print("═" * 60)

def kupiec_pof_test(returns_s, var_s, confidence):
    """
    Kupiec (1995) Proportion of Failures Test.
    LR statistic = -2 * log(L0/L1) ~ Chi²(1)
    """
    aligned    = pd.concat([returns_s, var_s], axis=1).dropna()
    actual_ret = aligned.iloc[:, 0]
    var_thresh = aligned.iloc[:, 1]

    breaches   = actual_ret < var_thresh
    n          = len(breaches)
    x          = breaches.sum()
    p_hat      = x / n
    p_expected = 1 - confidence

    if x == 0 or x == n:
        return {"breaches": x, "n": n, "actual_rate": p_hat,
                "expected_rate": p_expected, "LR": np.nan, "p_value": np.nan}

    # Likelihood ratio statistic
    LR = -2 * (
        x * np.log(p_expected / p_hat) +
        (n - x) * np.log((1 - p_expected) / (1 - p_hat))
    )
    p_value = 1 - stats.chi2.cdf(LR, df=1)

    return {"breaches": x, "n": n, "actual_rate": p_hat,
            "expected_rate": p_expected, "LR": LR, "p_value": p_value}

result = kupiec_pof_test(nifty_ret_aligned, var_garch_pct, CONFIDENCE)

print(f"\nTotal observations      : {result['n']}")
print(f"Expected breach rate    : {result['expected_rate']:.1%}  "
      f"({int(result['expected_rate']*result['n'])} expected breaches)")
print(f"Actual breach rate      : {result['actual_rate']:.1%}  "
      f"({result['breaches']} actual breaches)")
print(f"LR Test Statistic       : {result['LR']:.4f}")
print(f"p-value (Chi²)          : {result['p_value']:.4f}")

if result['p_value'] > 0.05:
    print(f"\n✓ Model PASSES backtesting (p > 0.05)")
    print(f"  Breach rate is statistically consistent with {CONFIDENCE*100:.0f}% VaR")
else:
    print(f"\n✗ Model FAILS backtesting (p < 0.05)")
    if result['actual_rate'] > result['expected_rate']:
        print(f"  Model UNDERESTIMATES risk — consider using t-distribution or EGARCH")
    else:
        print(f"  Model OVERESTIMATES risk — VaR is too conservative")


# ════════════════════════════════════════════════════════════════════
# CELL 16 — Multi-Index Volatility Correlation Heatmap
#
# Volatility correlations show how regimes are linked across sectors.
# High correlation between Nifty 50 and Nifty Bank = systemic risk.
# Low correlation with Nifty IT = different risk drivers.
# ════════════════════════════════════════════════════════════════════

print("═" * 60)
print("VOLATILITY CORRELATION ACROSS INDICES")
print("═" * 60)

# Build volatility DataFrame for all indices
vol_combined = pd.DataFrame({
    name: vol_data[name]["annual_vol"]
    for name in TICKERS.keys()
}).dropna()

# Return correlation matrix
corr_matrix = returns.corr()
vol_corr    = vol_combined.corr()

print("\nReturn Correlations:")
print(corr_matrix.round(4))
print("\nVolatility Correlations:")
print(vol_corr.round(4))

# Plot both heatmaps
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle("Correlation Matrices", fontsize=13, fontweight="bold")

for ax, matrix, title in zip(
    axes,
    [corr_matrix, vol_corr],
    ["Return Correlations", "Volatility Correlations"]
):
    im = ax.imshow(matrix, cmap="RdYlGn", vmin=-1, vmax=1, aspect="auto")
    ax.set_xticks(range(len(matrix.columns)))
    ax.set_yticks(range(len(matrix.columns)))
    ax.set_xticklabels(matrix.columns, rotation=30, ha="right", fontsize=9)
    ax.set_yticklabels(matrix.columns, fontsize=9)
    ax.set_title(title)
    for i in range(len(matrix)):
        for j in range(len(matrix.columns)):
            ax.text(j, i, f"{matrix.iloc[i, j]:.3f}",
                    ha="center", va="center", fontsize=9, fontweight="bold")
    plt.colorbar(im, ax=ax)

plt.tight_layout()
plt.savefig("05_correlation_heatmap.png", dpi=150, bbox_inches="tight")
plt.show()
print("✓ Chart saved as 05_correlation_heatmap.png")


# ════════════════════════════════════════════════════════════════════
# CELL 17 — Return Distribution by Regime
#
# This plot visually proves that regimes are meaningfully different.
# If the distributions overlap heavily, the HMM isn't discriminating well.
# Well-separated distributions = strong regime signal.
# ════════════════════════════════════════════════════════════════════

fig, axes = plt.subplots(1, N_REGIMES, figsize=(6 * N_REGIMES, 5))
fig.suptitle("Return Distribution by Regime — Nifty 50",
             fontsize=13, fontweight="bold")

if N_REGIMES == 1:
    axes = [axes]

for ax, (state_num, label) in zip(axes, label_map.items()):
    mask         = hidden_states == state_num
    state_ret    = nifty_ret.values[mask[:len(nifty_ret)]]
    color        = color_map[label]

    ax.hist(state_ret, bins=50, density=True, color=color,
            alpha=0.7, edgecolor="white", linewidth=0.3)

    # Overlay normal fit
    mu_s, std_s = state_ret.mean(), state_ret.std()
    x_range = np.linspace(state_ret.min(), state_ret.max(), 200)
    ax.plot(x_range, stats.norm.pdf(x_range, mu_s, std_s),
            "k--", linewidth=1.5, label="Normal fit")

    ax.axvline(mu_s, color="navy", linewidth=1.5, linestyle="-",
               label=f"Mean: {mu_s:.3f}%")
    ax.set_title(f"{label}\n{mask.sum()} days ({mask.mean()*100:.1f}%)")
    ax.set_xlabel("Daily Return (%)")
    ax.set_ylabel("Density")
    ax.legend(fontsize=8)
    ax.grid(alpha=0.3)
    ax.text(0.05, 0.95, f"σ = {std_s:.3f}%\nKurt = {kurtosis(state_ret, fisher=True):.2f}",
            transform=ax.transAxes, va="top", fontsize=9,
            bbox=dict(boxstyle="round", facecolor="white", alpha=0.7))

plt.tight_layout()
plt.savefig("06_regime_distributions.png", dpi=150, bbox_inches="tight")
plt.show()
print("✓ Chart saved as 06_regime_distributions.png")


# ════════════════════════════════════════════════════════════════════
# CELL 18 — Final Summary Dashboard (Plotly — Interactive)
# ════════════════════════════════════════════════════════════════════

# ════════════════════════════════════════════════════════════════════
# CELL 18 (FIXED) — Final Summary Dashboard (Plotly — Interactive)
# Fix: Pie chart needs specs=[..., {"type": "domain"}] in make_subplots
# ════════════════════════════════════════════════════════════════════

fig = make_subplots(
    rows=3, cols=2,
    subplot_titles=[
        "Nifty 50 Price (Regime Colored)",
        "GARCH Annualized Volatility (%)",
        "GARCH Dynamic VaR & CVaR (₹)",
        "Regime Distribution (% of Days)",   # ← this is the pie
        "Daily Returns vs VaR Threshold",
        "Multi-Index Volatility Comparison",
    ],
    # CRITICAL FIX: tell Plotly which cell is a pie (domain type)
    specs=[
        [{"type": "xy"},     {"type": "xy"}],
        [{"type": "xy"},     {"type": "domain"}],   # row 2 col 2 = pie
        [{"type": "xy"},     {"type": "xy"}],
    ],
    vertical_spacing=0.1,
    horizontal_spacing=0.08
)

# ── 1. Price with regime background ──────────────────────────────
price_aligned = prices["Nifty 50"].reindex(aligned.index)
fig.add_trace(go.Scatter(
    x=price_aligned.index, y=price_aligned,
    name="Nifty 50", line=dict(color="#3b82f6", width=1),
    showlegend=True
), row=1, col=1)

for state_num, label in label_map.items():
    mask = hidden_states == state_num
    idx  = aligned.index[mask]
    fig.add_trace(go.Scatter(
        x=idx, y=price_aligned.reindex(idx),
        mode="markers",
        marker=dict(color=color_map[label], size=2, opacity=0.5),
        name=label, legendgroup=label,
        showlegend=True
    ), row=1, col=1)

# ── 2. GARCH Volatility ───────────────────────────────────────────
annual_vol_all = vol_data["Nifty 50"]["annual_vol"]
fig.add_trace(go.Scatter(
    x=annual_vol_all.index, y=annual_vol_all,
    fill="tozeroy", name="Annual Vol",
    line=dict(color="#f59e0b"), fillcolor="rgba(245,158,11,0.25)",
    showlegend=False
), row=1, col=2)

# ── 3. GARCH VaR/CVaR ────────────────────────────────────────────
fig.add_trace(go.Scatter(
    x=cvar_garch_inr.index, y=cvar_garch_inr,
    fill="tozeroy", name="CVaR",
    line=dict(color="#ef4444"), fillcolor="rgba(239,68,68,0.2)",
    showlegend=False
), row=2, col=1)
fig.add_trace(go.Scatter(
    x=var_garch_inr.index, y=var_garch_inr,
    fill="tozeroy", name="VaR",
    line=dict(color="#f59e0b"), fillcolor="rgba(245,158,11,0.3)",
    showlegend=False
), row=2, col=1)

# ── 4. Regime pie chart (domain type — no row/col xy axes) ───────
regime_series  = pd.Series([label_map[s] for s in hidden_states])
regime_counts  = regime_series.value_counts()
fig.add_trace(go.Pie(
    labels=regime_counts.index,
    values=regime_counts.values,
    marker_colors=[color_map[l] for l in regime_counts.index],
    showlegend=False,
    textinfo="label+percent"
), row=2, col=2)   # Plotly now knows this cell is "domain" from specs

# ── 5. Returns vs VaR ────────────────────────────────────────────
ret_aligned2 = nifty_ret.reindex(var_garch_pct.index)
fig.add_trace(go.Bar(
    x=ret_aligned2.index, y=ret_aligned2,
    marker_color=np.where(ret_aligned2 >= 0, "#22c55e", "#ef4444"),
    name="Returns", showlegend=False, opacity=0.6
), row=3, col=1)
fig.add_trace(go.Scatter(
    x=var_garch_pct.index, y=var_garch_pct,
    line=dict(color="#f59e0b", width=1.2),
    name="VaR Threshold", showlegend=False
), row=3, col=1)

# ── 6. Multi-index volatility ─────────────────────────────────────
colors_multi = ["#3b82f6", "#ef4444", "#22c55e"]
for idx_name, col_ in zip(TICKERS.keys(), colors_multi):
    fig.add_trace(go.Scatter(
        x=vol_data[idx_name].index,
        y=vol_data[idx_name]["annual_vol"],
        name=idx_name, line=dict(color=col_, width=1),
        showlegend=True
    ), row=3, col=2)

# ── Layout ────────────────────────────────────────────────────────
fig.update_layout(
    height=900,
    title=dict(
        text=(f"<b>Multi-Asset Volatility Regime Dashboard</b><br>"
              f"<sub>GARCH(1,1) · HMM ({N_REGIMES} regimes) · "
              f"{CONFIDENCE*100:.0f}% VaR/CVaR · ₹{PORTFOLIO_VALUE:,.0f} Portfolio</sub>"),
        x=0.5, xanchor="center"
    ),
    hovermode="x unified",
    template="plotly_dark"
)

fig.show()
print("✓ Interactive dashboard rendered")
# ════════════════════════════════════════════════════════════════════
# CELL 19 — Download All Saved Charts from Colab
# ════════════════════════════════════════════════════════════════════

from google.colab import files
import os

chart_files = [f for f in os.listdir(".") if f.endswith(".png")]
print(f"Available charts: {chart_files}")
print("\nDownloading all charts...")

for f in chart_files:
    files.download(f)

print("✓ All charts downloaded")


# ════════════════════════════════════════════════════════════════════
# CELL 20 — Key Takeaways (Read After Running)
# ════════════════════════════════════════════════════════════════════

print("""
╔══════════════════════════════════════════════════════════════════╗
║   WHAT THIS PROJECT DEMONSTRATES (for your resume/interviews)   ║
╚══════════════════════════════════════════════════════════════════╝

1. GARCH MODELING
   ✓ Modeled time-varying volatility using GARCH(1,1)
   ✓ Interpreted α, β, persistence parameters
   ✓ Compared GARCH vs EGARCH vs GJR-GARCH using AIC/BIC
   ✓ Generated multi-step volatility forecasts

2. REGIME DETECTION
   ✓ Used HMM (Gaussian emissions) with Viterbi decoding
   ✓ Labeled regimes economically (Bull/Sideways/Bear)
   ✓ Analyzed transition matrix and expected regime durations
   ✓ Showed regimes have statistically different return distributions

3. RISK ESTIMATION
   ✓ Implemented Historical, Parametric, and GARCH-Dynamic VaR
   ✓ Computed CVaR (Expected Shortfall) — Basel III metric
   ✓ Backtested VaR using Kupiec's POF test (regulatory standard)
   ✓ Quantified risk in ₹ terms for a real portfolio

4. PORTFOLIO ANALYTICS
   ✓ Multi-index volatility correlation analysis
   ✓ Cross-sectoral risk comparison (Nifty 50, Bank, IT)

INTERVIEW TALKING POINTS:
  → "α+β near 1 means volatility shocks are highly persistent"
  → "GARCH-VaR adapts daily — historical VaR reacts slowly"
  → "CVaR is preferred over VaR because it captures tail severity"
  → "HMM captures regime switches that ARIMA cannot"
  → "Kupiec test ensures the model isn't just statistically fit
     but also practically valid for risk management"
""")
```
